```python
# TensorFlow and Keras
import tensorflow as tf
from tensorflow.keras import datasets, layers, models

# Numerical Operations
import numpy as np

# Data Visualization
import matplotlib.pyplot as plt

# Ignore Warnings
import warnings
warnings.filterwarnings('ignore')

# Check TensorFlow Version
print("TensorFlow Version:", tf.__version__)
```

```python
# Extract ZIP File
import zipfile
import os

zip_path = "/content/archive (2).zip"   # Replace with your file name

# Extract dataset

# Use zipfile module to extract .zip files
with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall()

print("Dataset Extracted Successfully!")
```

```python
from tensorflow.keras.datasets import cifar10

# Load Dataset
(x_train, y_train), (x_test, y_test) = cifar10.load_data()

```

```python
# Check Shape of Dataset

print("Training Images Shape:", x_train.shape)
print("Training Labels Shape:", y_train.shape)

print("Testing Images Shape:", x_test.shape)
print("Testing Labels Shape:", y_test.shape)


# Check Data Type

print("\nData Type of Images:", x_train.dtype)


# Check Minimum and Maximum Pixel Values

print("\nMinimum Pixel Value:", x_train.min())
print("Maximum Pixel Value:", x_train.max())


# Check Number of Classes

print("\nNumber of Classes:", len(np.unique(y_train)))


# Display First 10 Labels

print("\nFirst 10 Labels:")
print(y_train[:10].flatten())
```

```python
# Define CIFAR-10 Class Names

class_names = [
    'airplane',
    'automobile',
    'bird',
    'cat',
    'deer',
    'dog',
    'frog',
    'horse',
    'ship',
    'truck'
]

# Print Class Names

print("CIFAR-10 Classes:\n")

for index, class_name in enumerate(class_names):
    print(f"{index} --> {class_name}")
```

```python
# Display Sample Images from CIFAR-10 Dataset

plt.figure(figsize=(12, 12))

for i in range(25):

    # Create 5x5 Grid
    plt.subplot(5, 5, i + 1)

    # Remove Axis
    plt.xticks([])
    plt.yticks([])
    plt.grid(False)

    # Display Image
    plt.imshow(x_train[i])

    # Display Class Label
    plt.xlabel(class_names[y_train[i][0]])

# Show All Images
plt.show()
```

```python
# Normalize Pixel Values

x_train = x_train / 255.0
x_test = x_test / 255.0


# Check Pixel Range After Normalization

print("Minimum Pixel Value:", x_train.min())
print("Maximum Pixel Value:", x_train.max())
```

```python
# Build CNN Model

model = models.Sequential()

# =========================
# First Convolution Block
# =========================

model.add(layers.Conv2D(
    32,
    (3,3),
    padding='same',
    activation='relu',
    input_shape=(32,32,3)
))

model.add(layers.BatchNormalization())

model.add(layers.Conv2D(
    32,
    (3,3),
    padding='same',
    activation='relu'
))

model.add(layers.MaxPooling2D((2,2)))

model.add(layers.Dropout(0.25))


# =========================
# Second Convolution Block
# =========================

model.add(layers.Conv2D(
    64,
    (3,3),
    padding='same',
    activation='relu'
))

model.add(layers.BatchNormalization())

model.add(layers.Conv2D(
    64,
    (3,3),
    padding='same',
    activation='relu'
))

model.add(layers.MaxPooling2D((2,2)))

model.add(layers.Dropout(0.25))


# =========================
# Third Convolution Block
# =========================

model.add(layers.Conv2D(
    128,
    (3,3),
    padding='same',
    activation='relu'
))

model.add(layers.BatchNormalization())

model.add(layers.Conv2D(
    128,
    (3,3),
    padding='same',
    activation='relu'
))

model.add(layers.MaxPooling2D((2,2)))

model.add(layers.Dropout(0.25))


# =========================
# Fully Connected Layers
# =========================

model.add(layers.Flatten())

model.add(layers.Dense(
    256,
    activation='relu'
))

model.add(layers.Dropout(0.5))

# Output Layer
model.add(layers.Dense(
    10,
    activation='softmax'
))
```

```python
# Display Model Summary

model.summary()
```

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
# Train the CNN Model

history = model.fit(
    x_train,
    y_train,

    epochs=20,
    batch_size=64,

    validation_data=(x_test, y_test)
)
```

```python
# Evaluate CNN Model

test_loss, test_accuracy = model.evaluate(
    x_test,
    y_test
)

# Print Results

print("\nTest Accuracy:", test_accuracy)
print("Test Loss:", test_loss)
```

```python
# Plot Training and Validation Accuracy

plt.figure(figsize=(10, 5))

# Training Accuracy
plt.plot(
    history.history['accuracy'],
    label='Training Accuracy'
)

# Validation Accuracy
plt.plot(
    history.history['val_accuracy'],
    label='Validation Accuracy'
)

# Graph Title
plt.title('Training vs Validation Accuracy')

# Axis Labels
plt.xlabel('Epochs')
plt.ylabel('Accuracy')

# Legend
plt.legend()

# Show Graph
plt.show()
```

```python
# Plot Training and Validation Loss

plt.figure(figsize=(10, 5))

# Training Loss
plt.plot(
    history.history['loss'],
    label='Training Loss'
)

# Validation Loss
plt.plot(
    history.history['val_loss'],
    label='Validation Loss'
)

# Graph Title
plt.title('Training vs Validation Loss')

# Axis Labels
plt.xlabel('Epochs')
plt.ylabel('Loss')

# Legend
plt.legend()

# Show Graph
plt.show()
```
