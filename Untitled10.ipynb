```python
import tensorflow as tf
from tensorflow.keras.models import load_model
import numpy as np
import matplotlib.pyplot as plt
```

```python
loaded_model = load_model("cifar10_model.keras")

print("Model loaded successfully!")
```

```python
from tensorflow.keras.datasets import cifar10

(x_train, y_train), (x_test, y_test) = cifar10.load_data()

# Normalize
x_train = x_train / 255.0
x_test = x_test / 255.0
```

```python
class_names = [
    'airplane',
    'automobile',
    'bird',
    'cat',
    'deer',
    'dog',
    'frog',
    'horse',
    'ship',
    'truck'
]
```

```python
predictions = loaded_model.predict(x_test)

print("Prediction Shape:", predictions.shape)
```

```python


predicted_class = np.argmax(predictions[0])

actual_class = y_test[0][0]

print("Predicted:", class_names[predicted_class])
print("Actual   :", class_names[actual_class])
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 12))

for i in range(9):

    plt.subplot(3, 3, i + 1)

    plt.imshow(x_test[i])

    predicted_class = np.argmax(predictions[i])
    actual_class = y_test[i][0]

    color = "green" if predicted_class == actual_class else "red"

    plt.title(
        f"Pred: {class_names[predicted_class]}\n"
        f"Actual: {class_names[actual_class]}",
        color=color
    )

    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# Convert Predictions to Class Labels
y_pred = np.argmax(predictions, axis=1)

# Convert Actual Labels to 1D
y_true = y_test.flatten()

# Create Confusion Matrix
cm = confusion_matrix(y_true, y_pred)

# Plot
plt.figure(figsize=(12, 10))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=class_names,
    yticklabels=class_names
)

plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.title("CIFAR-10 Confusion Matrix")

plt.show()
```

```python
from sklearn.metrics import classification_report
import numpy as np

# Convert probabilities to class labels
y_pred = np.argmax(predictions, axis=1)

# Convert true labels to 1D
y_true = y_test.flatten()

# Generate Classification Report
report = classification_report(
    y_true,
    y_pred,
    target_names=class_names
)

print(report)
```
